```python
def classificar_por_regras(objeto, regras, ordem):
    texto = normalizar_texto(objeto)

    for enum in ordem:
        if enum in regras and re.search(regras[enum], texto):
            return CATEGORIAS_ESPECIFICAS[enum], enum

    return CATEGORIAS_ESPECIFICAS[22], 22
```

```python
def aplicar_classificacao(df_parte, regras, ordem):
    """
    Aplica um conjunto de regras a um DataFrame.
    Funciona mesmo se o DataFrame estiver vazio.
    """
    df_parte = df_parte.copy()

    if df_parte.empty:
        df_parte["categoriaEspecifica"] = pd.Series(dtype="object")
        df_parte["categoriaEspecificaEnum"] = pd.Series(dtype="int")
        return df_parte

    resultado = df_parte["objetoContrato"].apply(
        lambda x: pd.Series(classificar_por_regras(x, regras, ordem))
    )

    df_parte[["categoriaEspecifica", "categoriaEspecificaEnum"]] = resultado

    return df_parte


def classificar_direto(df_parte, enum):
    """
    Classifica diretamente um grupo inteiro em uma única categoria.
    """
    df_parte = df_parte.copy()
    df_parte["categoriaEspecifica"] = CATEGORIAS_ESPECIFICAS[enum]
    df_parte["categoriaEspecificaEnum"] = enum
    return df_parte


def reclassificar_outros_com_servicos(df_parte):
    """
    Pega os registros classificados como Outros em compras
    e tenta classificá-los novamente usando as regras de serviços.
    """
    df_parte = df_parte.copy()

    mask_outros = df_parte["categoriaEspecificaEnum"] == 22
    qtd_outros_antes = mask_outros.sum()

    if qtd_outros_antes == 0:
        print("Nenhum registro em Outros para reclassificar.")
        return df_parte

    resultado_servicos = df_parte.loc[mask_outros, "objetoContrato"].apply(
        lambda x: pd.Series(classificar_por_regras(x, REGRAS_SERVICOS, ORDEM_SERVICOS))
    )

    if resultado_servicos.empty:
        print("Nenhum resultado retornado pelas regras de serviços.")
        return df_parte

    # Reclassifica apenas os casos em que serviços encontrou algo diferente de Outros
    mask_reclassificados = resultado_servicos[1] != 22
    indices_reclassificados = resultado_servicos[mask_reclassificados].index

    df_parte.loc[indices_reclassificados, "categoriaEspecifica"] = resultado_servicos.loc[
        indices_reclassificados, 0
    ]

    df_parte.loc[indices_reclassificados, "categoriaEspecificaEnum"] = resultado_servicos.loc[
        indices_reclassificados, 1
    ]

    qtd_reclassificados = len(indices_reclassificados)
    qtd_outros_depois = (df_parte["categoriaEspecificaEnum"] == 22).sum()

    print(f"Outros antes: {qtd_outros_antes}")
    print(f"Reclassificados usando regras de serviços: {qtd_reclassificados}")
    print(f"Outros depois: {qtd_outros_depois}")

    return df_parte
```

```python
# Cópias para evitar SettingWithCopyWarning
cessao = cessao.copy()
compras = compras.copy()
informatica = informatica.copy()
internacional = internacional.copy()
locacao_imoveis = locacao_imoveis.copy()
mao_de_obra = mao_de_obra.copy()
obras = obras.copy()
servicos = servicos.copy()
servicos_engenharia = servicos_engenharia.copy()
servicos_saude = servicos_saude.copy()
alienacao = alienacao.copy()

# 1. Compras: classifica primeiro com regras de compras
compras = aplicar_classificacao(compras, REGRAS_COMPRAS, ORDEM_COMPRAS)

# 2. Compras que ficaram como Outros: tenta classificar com regras de serviços
compras = reclassificar_outros_com_servicos(compras)

# 3. Serviços: classifica com regras de serviços
servicos = aplicar_classificacao(servicos, REGRAS_SERVICOS, ORDEM_SERVICOS)

# 4. Internacional: usa compras + serviços, caso exista algum registro
REGRAS_GERAIS = {**REGRAS_COMPRAS, **REGRAS_SERVICOS}
ORDEM_GERAL = ORDEM_COMPRAS + ORDEM_SERVICOS

internacional = aplicar_classificacao(internacional, REGRAS_GERAIS, ORDEM_GERAL)

# 5. Categorias oficiais que já indicam diretamente o tipo
obras = classificar_direto(obras, 10)
servicos_engenharia = classificar_direto(servicos_engenharia, 10)
informatica = classificar_direto(informatica, 18)
locacao_imoveis = classificar_direto(locacao_imoveis, 15)
mao_de_obra = classificar_direto(mao_de_obra, 11)
servicos_saude = classificar_direto(servicos_saude, 17)
cessao = classificar_direto(cessao, 15)
alienacao = classificar_direto(alienacao, 21)
```

```python
df_classificado = pd.concat([
    cessao,
    compras,
    informatica,
    internacional,
    locacao_imoveis,
    mao_de_obra,
    obras,
    servicos,
    servicos_engenharia,
    servicos_saude,
    alienacao
], ignore_index=True)

df_classificado.to_csv(
    "../data/contratos_classificados_final.csv",
    index=False,
    encoding="utf-8-sig"
)
```
